```python
import pandas as pd

data = pd.read_csv("FinalDataset.csv", dtype={
    "Id": "string",
    "GZ12_Id": "string",
    "GZH_Id": "string",
    "GZC_Id": "string",
    "GZD_Id": "string",
    "Class": "string",
    'OriginalDatasetId': "string"})

sdss_bin = data[~data["GZ12_Id"].isna()] # turim SDSS duomenų
des_bin = data[~data["GZD_Id"].isna()] # turim DES duomenų
hubble_data = data[~data["GZC_Id"].isna() | ~data["GZH_Id"].isna()]

gzh_z_bin1 = hubble_data[hubble_data["z"] < 0.3]
gzh_z_bin2 = hubble_data[hubble_data["z"].between(0.3, 1.0, inclusive='left')]
gzh_z_bin3 = hubble_data[hubble_data["z"].between(1.0, 4.0, inclusive='both')]
```
