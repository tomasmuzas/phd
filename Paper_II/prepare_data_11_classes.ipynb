```sql
create table "AllGalaxies"
(
    "OriginalDatasetId"             text,
    "Class"                         text,
    "Id"                            uuid
);

create index original_id
    on "AllGalaxies" ("OriginalDatasetId");
```

````python
import pandas as pd

gz2_data = pd.read_csv("GZ2/gz2_hart16.csv")

# remove all artefacts
gz2_clean = gz2_data

gz2_clean = gz2_clean.rename(columns={
    "gz2_class": "Class",
    "dr7objid": "OriginalDatasetId"
    }).reset_index(drop=True)

gz2_clean["Class"] = gz2_clean["Class"].str.replace("[\+\(0-9\?].*", "", regex=True)

gz2_clean = gz2_clean[gz2_clean["Class"].isin(["Er", "Ei", "Ec", "Sa", "Sb", "Sc", "Sd", "SBa", "SBb", "SBc", "SBd"])]

gz2_clean[["OriginalDatasetId", "Class"]].to_csv("GZ2/gz2_11_classes.csv", index = False)
````
